Uploading PDF

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("RAG.pdf")
docs = loader.load()

print(docs[:2])
```

Chunking the PDF

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(docs)

print("Total chunks:", len(chunks))
print(chunks[0].page_content)
```

Creating Embeddings

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)
```

Creating Vector Database

```python
from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(
    documents=chunks,
    embedding=embedding,
    persist_directory="./chroma_db"
)

db.persist()
```

Retrieval (Search Engine)

```python
retriever = db.as_retriever()

query = "What is refund policy?"

results = retriever.invoke(query)

for r in results:
    print(r.page_content)
```

Answer Generation

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

query = "What is refund policy?"

docs = retriever.invoke(query)
context = " ".join([d.page_content for d in docs])

prompt = f"Context: {context}\n\nQuestion: {query}\nAnswer:"

response = generator(prompt, max_length=200, num_return_sequences=1)

print(response[0]["generated_text"])
```

LangGraph Implementation

Define State

```python
from typing import TypedDict

class GraphState(TypedDict):
    question: str
    context: str
    answer: str
    escalate: bool
```

Process Node (RAG logic)

```python
def process_node(state):
    query = state["question"]

    docs = retriever.invoke(query)

    # If nothing found → escalate
    if len(docs) == 0:
        return {
            "context": "",
            "answer": "No relevant data found",
            "escalate": True
        }

    context = " ".join([d.page_content for d in docs])

    prompt = f"Context: {context}\n\nQuestion: {query}\nAnswer:"

    response = generator(prompt, max_length=200)[0]["generated_text"]

    # 🔥 Simple decision logic
    if len(context) < 30 or "don't know" in response.lower():
        escalate = True
    else:
        escalate = False

    return {
        "context": context,
        "answer": response,
        "escalate": escalate
    }
```

HITL Node

```python
def hitl_node(state):
    print("\n⚠️ Escalated to Human Support")

    human_response = input("Enter human response: ")

    return {
        "answer": human_response,
        "escalate": False
    }
```

HITL Node

```python
def output_node(state):
    return state
```

Building Graph

```python
from langgraph.graph import StateGraph

workflow = StateGraph(GraphState)

# Add nodes
workflow.add_node("process", process_node)
workflow.add_node("hitl", hitl_node)
workflow.add_node("output", output_node)

# Entry point
workflow.set_entry_point("process")

# Routing logic
def route(state):
    if state["escalate"]:
        return "hitl"
    return "output"

# Conditional edges
workflow.add_conditional_edges("process", route)

# HITL → output
workflow.add_edge("hitl", "output")

# Compile
app = workflow.compile()
```

```python
result = app.invoke({
    "question": "What is refund policy?"
})

print("\n✅ Final Answer:")
print(result["answer"])
```
